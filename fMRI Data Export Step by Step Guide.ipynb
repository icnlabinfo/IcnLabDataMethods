***For system to turn on, Minerva would need to be present and switch on the system. If you are to do this personally, you would need e-learning, extra training and permissions.***

    1) Click on “Examination”.
    2) Click on Magnifying glass icon at the top left of the screen.
    3) Select the study folder from the table of folders shown on the screen.
    4) Hit the export button.
    5) In the window that pops up, you can select/deselect what you want, here you choose what gets saved in your USB drive.
    6) Once you are done, browse where to save the files:
    • “M” drive is for internal data (where data management pipeline is, where the data cluster is), if data gets transferred there, let management know IMMEDIATELY because confidential data will be backed up when it should have been deleted.
    • “Med System” is the computer connected to the monitor you are currently working on.
    • “D” is the USB drive you need to choose.
    7) Hit okay.
    8) For image conversion drop down menu choose “Interoperability”. DICOMS for interoperability will save every slice as an individual DICOMS image, “Enhanced” will give every slice as a whole. MINDSET Project can use Enhanced.
    9) In the “Export Data” window don’t select anything, you can only change conversion to “Interoperability” or leave on “Enhanced”, if the facility you send the data to cannot work with “Enhanced” then they will let you know and you can go back to save the files with “Interoperability” conversion.
    10) Hit Export.
    11) Double click on CD icon that shows up in the bottom right of the toolbar and double click.
    12) Once you see that the process is complete, it will show 100% on the screen. You are now safe to exit from that window.
    13) Remove USB from computer and move to a internet connected computer that is UNLV property.
    14) Open the MINDSET account link and sign in.
    15) Select choose file on the screen.
    16) Go to your USB Drive and select the folder you just moved into the USB.
    17) Click upload.
    18) Wait for confirmation email that they have received the files.
    19) You are now required to delete the files from the USB. Use Shift key on the keyboard + delete key on the keyboard to permanently delete the files.
    20) Let Minerva know so she can delete the files from the monitor.
